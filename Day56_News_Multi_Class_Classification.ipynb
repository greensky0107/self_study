# Revert dataset into text

```python
# word_index가 vocabulary
word_index = reuters.get_word_index(path="reuters_word_index.json")
print('=3')
```

```python
# the는 정수 1에 mapping
word_index['the']
```

```python
# advise는 정수 5233에 mapping
word_index['advise']
```

```python
# 0번(<pad>), 1번(<sos>), 2번(<unk>)은 자연어 처리에 쓰이는 특별한 토큰들을 위한 번호
# 따라서 index를 word로 돌릴 때는 각 번호에 3을 더해주어야한다.
index_to_word = { index+3 : word for word, index in word_index.items() }
print('=3')
```

```python
print(index_to_word[4])
print(index_to_word[5236])
```

```python
# index_to_word에 숫자 0은 <pad>, 숫자 1은 <sos>, 숫자 2는 <unk>를 넣어줍니다.
# # UNK=unknown=out-of-vocabulary 정수번호가 10,000이 넘는 단어는 UNK라는 tocken으로 변환  (num_words=10000)
for index, token in enumerate(("<pad>", "<sos>", "<unk>")):
  index_to_word[index]=token
print('=3')
```

```python
print(' '.join([index_to_word[index] for index in x_train[0]]))
# 0번 뉴스가 index 에서 word로 바뀜
```

```python
# 로이터 뉴스 데이터 중 [4, 587, 23, 133, 6, 30, 515]인 정수 시퀀스가 있습니다. 현재 우리가 가진 index_word를 사용하여 이 정수 시퀀스를 텍스트 시퀀스로 변환해보세요.

indexs = [4, 587, 23, 133, 6, 30, 515]

result = ' '.join([index_to_word[index] for index in indexs])
result
```

```python
# num_words=10000으로 정했으니, 12000은 unk로 tocken화.
indexs = [4, 12000, 23, 133, 6, 30, 515]

result = ' '.join([index_to_word[index] for index in indexs])
result
```

```python
# 전체 news를 문장(텍스트)으로 다시 돌림
decoded = []
for i in range(len(x_train)):
    t = ' '.join([index_to_word[index] for index in x_train[i]])
    decoded.append(t)

x_train = decoded
print(len(x_train))
```

```python
#test데이터도 변환해주세요!!!
decoded_test = []
for i in range(len(x_test)):
    t = ' '.join([index_to_word[index] for index in x_test[i]])
    decoded_test.append(t)

x_test = decoded_test
print(len(x_test))
```

# Vectorization

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfTransformer
print('=3')
```

```python
# train dataset을 DTM(Document Term Matrix) 방법으로 vectorization (news는 8982개,  word는 9670개)
dtmvector = CountVectorizer()
x_train_dtm = dtmvector.fit_transform(x_train)
print(x_train_dtm.shape)
```

```python
# DTM으로 변환된 dataset을 TF-IDF(Term Frequency-Inverse Document Frequency) 방법으로 vectorization
tfidf_transformer = TfidfTransformer()
tfidfv = tfidf_transformer.fit_transform(x_train_dtm)
print(tfidfv.shape)
```

# Naive Bayes Classification  (Evaluation Metrics - Accuracy)

```python
from sklearn.naive_bayes import MultinomialNB #다항분포 나이브 베이즈 모델  Bayes 정리
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.naive_bayes import ComplementNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.ensemble import VotingClassifier
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score #정확도 계산

print('=3')
```

```python
model = MultinomialNB()
model.fit(tfidfv, y_train)
```

```python
x_test_dtm = dtmvector.transform(x_test) #테스트 데이터를 DTM으로 변환
tfidfv_test = tfidf_transformer.transform(x_test_dtm) #DTM을 TF-IDF 행렬로 변환

predicted = model.predict(tfidfv_test) #테스트 데이터에 대한 예측
print("정확도:", accuracy_score(y_test, predicted)) #예측값과 실제값 비교
```

# Naive Bayes Classification (Evaluation Metrics - F-1 Score, Confusion Matrix)

```python
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

print('=3')
```

```python
print(classification_report(y_test, model.predict(tfidfv_test), zero_division=0))
# classification_report() 함수는 각각의 class를 positive class로 보았을 때의 정밀도, 재현율, F1점수를 각각 구하고 그 평균값으로 전체 모델의 성능을 평가.
```

macro: 단순평균
weighted: 각 클래스에 속하는 표본의 개수로 가중평균
accuracy: 정확도. 전체 학습 데이터의 개수에서 클래스를 정확하게 맞춘 개수의 비율

```python
def graph_confusion_matrix(model, x_test, y_test):#, classes_name):
  df_cm = pd.DataFrame(confusion_matrix(y_test, model.predict(x_test)))#, index=classes_name, columns=classes_name)
  fig = plt.figure(figsize=(12,12))
  heatmap = sns.heatmap(df_cm, annot=True, fmt="d")
  heatmap.yaxis.set_ticklabels(heatmap.yaxis.get_ticklabels(), rotation=0, ha='right', fontsize=12)
  heatmap.xaxis.set_ticklabels(heatmap.xaxis.get_ticklabels(), rotation=45, ha='right', fontsize=12)
  plt.ylabel('label')
  plt.xlabel('predicted value')

print('=3')
```

```python
graph_confusion_matrix(model, tfidfv_test, y_test)
```

# Complement Naive Bayes Classifier(CNB)


데이터가 불균형할 경우를 대비해 Naive Bayes Classifier를 보완한 것이 Complement Naive Bayes Classifier. 데이터의 불균형을 고려하여 가중치를 부여하는 특징을 가지고 있음.  일반적으로 Complement Naive Bayes Classifier가 Naive Bayes Classifier(MultinomialNB)보다 성능이 일반적으로 더 좋음.

```python
cb = ComplementNB()
cb.fit(tfidfv, y_train)
```

```python
predicted = cb.predict(tfidfv_test) #테스트 데이터에 대한 예측
print("정확도:", accuracy_score(y_test, predicted)) #예측값과 실제값 비교
```

# Voting

```python
# 26분정도 소요됨
log_clf = LogisticRegression(penalty='l2', solver='lbfgs', max_iter=1000)
nb_clf = ComplementNB()
gb_clf = GradientBoostingClassifier()

# Voting Classifier 선언
voting_classifier = VotingClassifier(
    estimators=[
        ('lr', log_clf),
        ('nb', nb_clf),
        ('gb', gb_clf)
    ],
    voting='soft'
)
voting_classifier.fit(tfidfv, y_train)
```

```python
predicted = voting_classifier.predict(tfidfv_test) #테스트 데이터에 대한 예측
print("정확도:", accuracy_score(y_test, predicted)) #예측값과 실제값 비교
```
