# 1.Cosine Similarity

```python
import numpy as np

t1 = np.array([1, 1, 1])
t2 = np.array([2, 0, 1])
```

```python
from numpy import dot
from numpy.linalg import norm
def cos_sim(A, B):
	return dot(A, B)/(norm(A)*norm(B))
```

```python
cos_sim(t1, t2)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
t1 = np.array([[1, 1, 1]])
t2 = np.array([[2, 0, 1]])
cosine_similarity(t1,t2)
```

```python
# Q. 아래 t1, t2 리스트의 요소 값을 조정해서 코사인 유사도 값이 0이 되게 만들어보세요.
t1 = np.array([[1, 1, 1]])
t2 = np.array([[-1, -1, 2]])
cosine_similarity(t1,t2)
```

```python
# Q. 코사인 유사도 값이 -1이 되게도 만들어보세요.
t1 = np.array([[1, 2, 3]])
t2 = np.array([[-1, -2, -3]]) # Reshape t2 to be a 2D array
cosine_similarity(t1,t2)
```

# 2. Contents based filtering

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print('⏩⏩')
```

```python
import os
# Removed os.getenv('HOME') which was adding '/root'
csv_path = '/content/sample_data/movie_dataset.csv'
df = pd.read_csv(csv_path)
df.head()
```

```python
df.columns
```

```python
features = ['keywords','cast','genres','director']
features
```

```python
def combine_features(row):
    return row['keywords']+" "+row['cast']+" "+row['genres']+" "+row['director']

combine_features(df[:5])
```

```python
#combined_features를 새로운 column으로 생성함.
for feature in features:
    df[feature] = df[feature].fillna('')

df["combined_features"] = df.apply(combine_features,axis=1)
df["combined_features"]
```

```python
# vectorize하고 cosine_similarity 구함
cv = CountVectorizer()
count_matrix = cv.fit_transform(df["combined_features"])
print(type(count_matrix))
print(count_matrix.shape)
print(count_matrix)

# 1번째 row (0,3115) 1 은 3116번째 단어가 1번 출현한다는 뜻
# 14845개의 단어가 존재하는데, 이 단어들을 범주형으로 보고
# 그 단어의 출현 빈도만을 표시한 Matrix가 매우 Sparse하기 때문에 공간을 절약할 수 있는 형태로 표현한 것
```

```python
# vectorized 된 4,803 개 영화의 cosine similarity 구하기
cosine_sim = cosine_similarity(count_matrix)
print(cosine_sim)
print(cosine_sim.shape)
```

```python
# cosine similarity를 이용해서 영화추천하기
def get_title_from_index(index):
    return df[df.index == index]["title"].values[0]
def get_index_from_title(title):
    return df[df.title == title]["index"].values[0]

movie_user_likes = "Avatar"
movie_index = get_index_from_title(movie_user_likes)
similar_movies = list(enumerate(cosine_sim[movie_index])) # cosine similarity

sorted_similar_movies = sorted(similar_movies,key=lambda x:x[1],reverse=True)[1:]

i=0
print(movie_user_likes+"와 비슷한 영화 3편은 "+"\n")
for item in sorted_similar_movies:
    print(get_title_from_index(item[0]))
    i=i+1
    if i==3:
        break
```

# 3. Collaborative filtering

```python
import numpy as np
from numpy.linalg import svd
```

```python
np.random.seed(30)
A = np.random.randint(0, 100, size=(4, 4))
A
```

```python
svd(A)
```

```python
U, Sigma, VT = svd(A)

print('U matrix: {}\n'.format(U.shape),U)
print('Sigma: {}\n'.format(Sigma.shape),Sigma)
print('V Transpose matrix: {}\n'.format(VT.shape),VT)
```

```python
Sigma_mat = np.diag(Sigma)

A_ = np.dot(np.dot(U, Sigma_mat), VT)
A_
```
